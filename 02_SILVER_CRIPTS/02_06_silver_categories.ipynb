### TRANSFORMATION WORKFLOW

1. TRIM SPACES
2. RENAME COLUMNS 
3. WRITE INTO SILVER

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType
```

```python
# 0) LOAD DATA & READ FROM BRONZE
df_erpcat = spark.table("acdproj.bronze.erp_px_cat_g1v2")
```

```python
# 1) TRIM SPACES
for field in df_erpcat.schema.fields:
    if isinstance(field.dataType, StringType):
        df_erpcat = df_erpcat.withColumn(field.name, F.trim(F.col(field.name)))
```

```python
# 2) RENAME COLUMNS
RENAME_MAP_CAT = {
    "ID": "category_id",
    "CAT": "category",
    "SUBCAT": "subcategory",
    "MAINTENANCE": "maintenance"
}
for old_name, new_name in RENAME_MAP_CAT.items():
    df_erpcat = df_erpcat.withColumnRenamed(old_name, new_name)
```

```python
# 3) WRITE INTO SILVER
df_erpcat.write.mode("overwrite").saveAsTable("acdproj.silver.erp_categories")
```
