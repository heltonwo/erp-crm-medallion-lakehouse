# Bronze 
### Methodology - Extract and Load
1. Read (From Source)
2. Load/ Write (From Raw Source into Bronze Table)

### TRANSFORMATION WORKFLOW 
1. EXTRACT AND LOAD... ALL RAW

```python
# 1) LOAD ALL CRM (# 1 Load CRM - Customer information table, # 2) Load CRM - Product information table, # 3) Load CRM - Sales information table)

folder_path = "/Volumes/dbacademy/sc_raw/raw_sources/source_crm/"
files = dbutils.fs.ls(folder_path)

for file in files:
    if file.name.endswith(".csv"):
        table_name = file.name.replace(".csv", "").lower()
        full_table_name = f"acdproj.bronze.crm_{table_name}"

        print(f"Loading {file.path} -> {full_table_name}")

        df = (
            spark.read
            .option("header", "true")
            .option("inferSchema", "true")
            .csv(file.path)
        )

        df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(full_table_name)

print("Done loading all CRM bronze tables.")
```

```python
# 2) LOAD ALL ERP (# 1 Load ERP - Customer/ID information table, # 2) Load ERP - Localization information table, # 3) Load CRM - Category information table)

folder_path = "/Volumes/dbacademy/sc_raw/raw_sources/source_erp"
files = dbutils.fs.ls(folder_path)

for file in files:
    if file.name.endswith(".csv"):
        table_name = file.name.replace(".csv", "").lower()
        full_table_name = f"acdproj.bronze.erp_{table_name}"

        print(f"Loading {file.path} -> {full_table_name}")

        df = (
            spark.read
            .option("header", "true")
            .option("inferSchema", "true")
            .csv(file.path)
        )

        df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(full_table_name)

print("Done loading all ERP bronze tables.")
```
