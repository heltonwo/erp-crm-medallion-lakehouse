### ENVIRONMENT SETUP

1. CREATE CATALOG (IF NOT EXISTS)
2. CREATE SCHEMAS: BRONZE, SILVER, GOLD

**Run this notebook ONCE** when setting up the project from scratch, or after a full
reset. It is NOT part of the regular bronze → silver → gold pipeline run — running
it again is safe (idempotent), but unnecessary once the environment already exists.

```python
# 1) DEFINE SCHEMAS TO CREATE
schemas = ["bronze", "silver", "gold"]

# 2) CREATE CATALOG AND SET IT AS ACTIVE
spark.sql("CREATE CATALOG IF NOT EXISTS acdproj")
spark.sql("USE CATALOG acdproj")

# 3) CREATE EACH SCHEMA (idempotent — safe to re-run, never drops existing data)
for schema in schemas:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
    print(f"Schema '{schema}' ready.")
```
